Assuming a dataset with 1d, lat and long for dbscan

```python
#Prepare coordinates for DBSCAN
coords = df[['lat', 'long']].values  # Extract lat, long as numpy array
coords_radians = np.radians(coords)  # Convert to radians for Haversine distance
```

DBSCAN clusters points based on density, using eps (maximum distance for neighbors) and min_samples (minimum points for a cluster).

```python
# Step 3: Apply DBSCAN
# eps = 1 km (convert to radians: 1/6371), min_samples = 10
earth_radius = 6371  # Earth's radius in km
eps_km = 1  # Cluster radius in kilometers
eps_radians = eps_km / earth_radius
min_samples = 10

db = DBSCAN(eps=eps_radians, min_samples=min_samples, metric='haversine', n_jobs=-1)
df['cluster'] = db.fit_predict(coords_radians)

# Step 4: Summarize results
n_clusters = len(set(df['cluster']) - {-1})  # Exclude noise (-1)
n_noise = len(df[df['cluster'] == -1])
print(f"Found {n_clusters} clusters and {n_noise} noise points")

# Step 5: Save the clustered DataFrame
df.to_csv('clustered_data.csv', index=False)
print("Clustered DataFrame saved as 'clustered_data.csv'")
```
